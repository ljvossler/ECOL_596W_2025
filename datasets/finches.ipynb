```python
import pandas as pd
```

```python
df = pd.read_csv('finch.csv', encoding='latin-1')
```

```python
df
```

```python
prop_dct = {}
site_types = set(df['type_of_site'])
for site in site_types:
    total = 0
    infected = 0
    for row in df.index:
        if df.loc[row, 'type_of_site'] == site:
            total +=1
        if df.loc[row, 'pox_IUR'] == 'I':
            infected += 1
    prop_dct[site] = [infected / total]
```

```python
infected
```

```python
prop_dct
```

```python
prop_df = pd.DataFrame(prop_dct)
prop_df
```
